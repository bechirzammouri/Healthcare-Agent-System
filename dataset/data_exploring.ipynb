```python
import pandas as pd 
```

```python
medications = pd.read_csv("sampled_medications.csv")
conditions = pd.read_csv("sampled_conditions.csv")
```

```python
medications.columns.to_list()
```

```python
conditions.columns.to_list()
```

```python
len(medications["CODE"] )
```

```python
len(medications["CODE"].unique())
```

```python
medication_codes = set(medications['REASONCODE'].dropna().astype(int).unique())
condition_codes = set(conditions['CODE'].astype(int).unique())

common_codes = medication_codes.intersection(condition_codes)

print(f"Medications has {len(medication_codes)} unique codes")
print(f"Conditions has {len(condition_codes)} unique codes")
print(f"Common codes: {len(common_codes)}")
print(f"Common codes list: {sorted(common_codes)}")
```

```python
careplans = pd.read_csv("sampled_careplans.csv")
```

```python
careplans_reasonCode = set(careplans['REASONCODE'].dropna().astype(int).unique())
condition_codes = set(conditions['CODE'].astype(int).unique())

common_codes = careplans_reasonCode.intersection(condition_codes)

print(f"careplans has {len(careplans_reasonCode)} unique codes")
print(f"Conditions has {len(condition_codes)} unique codes")
print(f"Common codes: {len(common_codes)}")
print(f"Common codes list: {sorted(common_codes)}")
```
