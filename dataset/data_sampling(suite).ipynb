```python
import pandas as pd 

data_path = "SyntheticMass_Data"
target_data_path = "sampled_data"
patients = pd.read_csv(target_data_path + "/sampled_patients.csv")
```

### sampling procedures

```python
procedures = pd.read_csv(f"{data_path}/procedures.csv")
procedures.head()
```

```python
def filter(dataset):
    filter = dataset[dataset['PATIENT'].isin(patients['ID'])]
    print(f"Filtered data count related to {len(patients)} patients: {len(filter)}")
    return filter
```

```python
sampled_procedures = filter(procedures)
```

```python
sampled_procedures.to_csv(target_data_path + "/sampled_procedures.csv", index=False)
del sampled_procedures
```

```python
del procedures
```

### sampling observations

```python
# Process observations in chunks to avoid memory issues
chunk_size = 100000  # Adjust based on your available memory
first_chunk = True

for chunk in pd.read_csv(f"{data_path}/observations.csv", chunksize=chunk_size):
    # Filter each chunk for our sampled patients
    filtered_chunk = chunk[chunk['PATIENT'].isin(patients['ID'])]
    
    # Append to CSV (write header only for first chunk)
    filtered_chunk.to_csv(target_data_path + "/sampled_observations.csv", 
                         mode='a' if not first_chunk else 'w',
                         header=first_chunk,
                         index=False)
    
    if first_chunk:
        first_chunk = False
        print(f"Started processing observations in chunks of {chunk_size}")
    
    # Clean up to free memory
    del chunk, filtered_chunk

print(f"Finished processing observations")
# Verify the result
sampled_observations = pd.read_csv("sampled_observations.csv")
print(f"Filtered observations count related to {len(patients)} patients: {len(sampled_observations)}")
del sampled_observations
```
