```python
medication_codes = set(medications['REASONCODE'].dropna().astype(int).unique())
condition_codes = set(conditions['CODE'].astype(int).unique())

common_codes = medication_codes.intersection(condition_codes)

print(f"Medications has {len(medication_codes)} unique codes")
print(f"Conditions has {len(condition_codes)} unique codes")
print(f"Common codes: {len(common_codes)}")
print(f"Common codes list: {sorted(common_codes)}")
```

checking if encouter date is the same immunization date

```python
immunizations = pd.read_csv("sampled_data/sampled_immunizations.csv")
encounters = pd.read_csv("sampled_data/sampled_encounters.csv")
```

```python
# Merge immunizations with encounters on ENCOUNTER ID
merged = immunizations.merge(encounters, left_on='ENCOUNTER', right_on='ID', how='left')

# Convert DATE columns to datetime for comparison
merged['DATE_x'] = pd.to_datetime(merged['DATE_x'])
merged['DATE_y'] = pd.to_datetime(merged['DATE_y'])

# Check if dates match
merged['dates_match'] = merged['DATE_x'] == merged['DATE_y']

# Display results
print(f"Total records: {len(merged)}")
print(f"Matching dates: {merged['dates_match'].sum()}")
print(f"Non-matching dates: {(~merged['dates_match']).sum()}")
print("\nDetails:")
print(merged[['DATE_x', 'DATE_y', 'dates_match']])
```
